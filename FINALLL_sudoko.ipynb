## Choosing Level of the Game

1-Initializes Pygame and sets up the window and font, which are essential for displaying anything on the screen and interacting with the user

2-Handles drawing text on the screen. The display_prompt function shows both the input prompt and error messages, keeping the user informed about what they need to do and any mistakes they make.

3-Handles user interactions by checking key presses and updating the input accordingly. It ensures that the input is valid, processes special keys like Backspace and Enter

```python
def get_difficulty():
    pygame.init() #starts up the Pygame library
    win = pygame.display.set_mode((WIDTH, WIDTH))
    pygame.display.set_caption("Select Difficulty")
    myfont = pygame.font.SysFont('Comic Sans MS', 35)
    
    def display_prompt(message, error=False): #dy function 3lshan tbainlna el message fy el awel
        win.fill(background_color)
        difficulty_prompt = myfont.render("Enter difficulty (easy, medium, hard):", True, (0, 0, 0))
        win.blit(difficulty_prompt, (10, 10)) #position of the text
        if error:  #dy 3lshan le dakhlna invalid level
            error_message = myfont.render(message, True, (255, 0, 0))
            win.blit(error_message, (10, 60))
        pygame.display.update()
    
    display_prompt("") #shows the initial message asking for difficulty without any error

    difficulty = "" #sets up a blank space to keep track of what the user is typing
    while True: 
        for event in pygame.event.get(): #function to check every action
            if event.type == pygame.KEYDOWN: #checks if a key on the keyboard has been pressed
                if event.key == pygame.K_RETURN: #for enter key
                    if difficulty in ["easy", "medium", "hard"]:
                        pygame.quit() #close el window bta3t choosing the difficulty
                        return difficulty
                    else:
                        # Reset input and show error message
                        difficulty = ""
                        display_prompt("Invalid input. Please enter (easy, medium, hard):", error=True)
                elif event.key == pygame.K_BACKSPACE: 
                    difficulty = difficulty[:-1]  #removes the last letter from what the user typed
                    display_prompt("")  # Clear input text display
                    difficulty_text = myfont.render(difficulty, True, (0, 0, 0)) #show what user has typed
                    win.blit(difficulty_text, (10, 50))
                    pygame.display.update()
                elif event.key <= 127:  # checks if the key pressed is a normal letter or number.
                    difficulty += chr(event.key)
                    display_prompt("")  # Clear input text display
                    difficulty_text = myfont.render(difficulty, True, (0, 0, 0)) #show updated text
                    win.blit(difficulty_text, (10, 50))
                    pygame.display.update()
            elif event.type == pygame.QUIT: #This checks if the user closes the window.
                pygame.quit()
                sys.exit()
```

## Validation

function is designed to check whether a number can be placed in a specific position on a Sudoku board without violating Sudoku rules

```python
def is_valid(board, row, col, num):
    if num in board[row]:
        return False
    for r in range(9):
        if board[r][col] == num:  #Iterates through each row in the specified column (khlinha kda msh zay el row 3lshan lma solve function nendhlha)
            return False
    start_row, start_col = 3 * (row // 3), 3 * (col // 3) # Calculates the starting row and starting column of the 3x3 sub-grid
    for r in range(start_row, start_row + 3): #iterate for each cell in the 3x3 grid
        for c in range(start_col, start_col + 3):
            if board[r][c] == num:
                return False
    return True
```

## Error Message

The function temporarily displays a red "Invalid move!" error message on the screen when the player makes an illegal move. The message stays visible for a short time, then the game screen is redrawn with the grid, numbers, and buttons back in place.

```python
def show_error_message(win, myfont):
    error_message = myfont.render("Invalid move!", True, (255, 0, 0))
    win.fill(background_color)
    pygame.draw.rect(win, (255, 255, 255), (0, 0, WIDTH, WIDTH))
    win.blit(error_message, (WIDTH // 2 - 50, WIDTH // 2 - 20)) #to place the message in the center of the window
    pygame.display.update()
    pygame.time.delay(error_message_duration)
    #after the delay the window is filled again with 
    win.fill(background_color)
    draw_grid(win)
    draw_numbers(win)
    draw_buttons(win)
    pygame.display.update()
```

## Solve a Sudoku puzzle

### Recursive Approach:
The function uses recursion to explore possible numbers for each empty cell.

### Backtracking: 
It backtracks by resetting cells to try different configurations if the current setup does not lead to a solution.

### Base Case: 
The base case is when there are no empty cells left, indicating that the board is completely and correctly filled.

```python
def solve_sudoku(board):
    empty = find_empty_location(board)
    if not empty:
        return True
    row, col = empty
    for num in range(1, 10):
        if is_valid(board, row, col, num):
            board[row][col] = num
            if solve_sudoku(board): #it means the puzzle can be solved with this number, so return True
                return True
            board[row][col] = 0 # lw wala rakm true fa current borad is not solvable
    return False
```

```python
def find_empty_location(board):
    for i in range(9):
        for j in range(9):
            if board[i][j] == 0:
                return (i, j)
    return None
```
