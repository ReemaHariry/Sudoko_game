```python
import pygame
import sys

WIDTH = 650
background_color = (251,247,245)
original_grid_element_color = (52, 31, 151)
button_color = (0, 255, 0)
button_hover_color = (0, 200, 0)
buffer = 5
error_message_duration = 1000  # Duration in milliseconds

# Define Sudoku boards for different difficulties
easy_board = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9]
]

medium_board = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9]

]

hard_board = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9]
]

def select_board(difficulty):
    if difficulty == "easy":
        return easy_board
    elif difficulty == "medium":
        return medium_board
    elif difficulty == "hard":
        return hard_board
    else:
        raise ValueError("Unknown difficulty level")

def get_difficulty():
    pygame.init()
    win = pygame.display.set_mode((WIDTH, WIDTH))
    pygame.display.set_caption("Select Difficulty")
    win.fill(background_color)
    myfont = pygame.font.SysFont('Comic Sans MS', 35)
    
    difficulty_prompt = myfont.render("Enter difficulty (easy, medium, hard):", True, (0, 0, 0))
    win.blit(difficulty_prompt, (10, 10))
    
    pygame.display.update()
    
    difficulty = ""
    while True:
        for event in pygame.event.get():
            if event.type == pygame.KEYDOWN:
                if event.key == pygame.K_RETURN:
                    if difficulty in ["easy", "medium", "hard"]:
                        pygame.quit()
                        return difficulty
                    else:
                        difficulty_prompt = myfont.render("Invalid input. Please enter (easy, medium, hard):", True, (255, 0, 0))
                        win.blit(difficulty_prompt, (10, 60))
                        pygame.display.update()
                elif event.key == pygame.K_BACKSPACE:
                    difficulty = difficulty[:-1]
                elif event.key <= 127:  # ASCII printable characters
                    difficulty += chr(event.key)
                    difficulty_text = myfont.render(difficulty, True, (0, 0, 0))
                    win.fill(background_color)
                    win.blit(difficulty_prompt, (10, 10))
                    win.blit(difficulty_text, (10, 50))
                    pygame.display.update()
            elif event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()

def is_valid(board, row, col, num):
    if num in board[row]:
        return False
    for r in range(9):
        if board[r][col] == num:
            return False
    start_row, start_col = 3 * (row // 3), 3 * (col // 3)
    for r in range(start_row, start_row + 3):
        for c in range(start_col, start_col + 3):
            if board[r][c] == num:
                return False
    return True

def insert(win, position):
    i, j = position[1], position[0]
    myfont = pygame.font.SysFont('Comic Sans MS', 35)
    
    while True:
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                return
            if event.type == pygame.KEYDOWN:
                if(grid_original[i-1][j-1] != 0):
                    return
                if(event.key == 48): # checking with 0
                    grid[i-1][j-1] = event.key - 48
                    pygame.draw.rect(win, background_color, (position[0]*50 + buffer, position[1]*50 + buffer, 50 - 2*buffer , 50 - 2*buffer))
                    pygame.display.update()
                    return
                if(0 < event.key - 48 < 10):
                    num = event.key - 48
                    if is_valid(grid, i-1, j-1, num):
                        pygame.draw.rect(win, background_color, (position[0]*50 + buffer, position[1]*50 + buffer, 50 - 2*buffer , 50 - 2*buffer))
                        value = myfont.render(str(num), True, (0,0,0))
                        win.blit(value, (position[0]*50 + 15, position[1]*50))
                        grid[i-1][j-1] = num
                        pygame.display.update()
                        return
                    else:
                        show_error_message(win, myfont)
                        return

def show_error_message(win, myfont):
    error_message = myfont.render("Invalid move!", True, (255, 0, 0))
    win.fill(background_color)
    pygame.draw.rect(win, (255, 255, 255), (0, 0, WIDTH, WIDTH))
    win.blit(error_message, (WIDTH // 2 - 50, WIDTH // 2 - 20))
    pygame.display.update()
    pygame.time.delay(error_message_duration)
    win.fill(background_color)
    draw_grid(win)
    draw_numbers(win)
    draw_buttons(win)
    pygame.display.update()

def draw_grid(win):
    for i in range(0, 10):
        if(i % 3 == 0):
            pygame.draw.line(win, (0, 0, 0), (50 + 50*i, 50), (50 + 50*i , 500 ), 4 )
            pygame.draw.line(win, (0, 0, 0), (50, 50 + 50*i), (500, 50 + 50*i), 4 )
        pygame.draw.line(win, (0, 0, 0), (50 + 50*i, 50), (50 + 50*i , 500 ), 2 )
        pygame.draw.line(win, (0, 0, 0), (50, 50 + 50*i), (500, 50 + 50*i), 2 )

def draw_numbers(win):
    myfont = pygame.font.SysFont('Comic Sans MS', 35)
    for i in range(len(grid)):
        for j in range(len(grid[0])):
            if(0 < grid[i][j] < 10):
                value = myfont.render(str(grid[i][j]), True, original_grid_element_color)
                win.blit(value, ((j+1)*50 + 15, (i+1)*50 )) 

def create_button(win, text, position, size):
    font = pygame.font.SysFont('Comic Sans MS', 30)
    button_rect = pygame.Rect(position, size)
    pygame.draw.rect(win, button_color, button_rect)
    button_text = font.render(text, True, (0, 0, 0))
    win.blit(button_text, (position[0] + 10, position[1] + 10))
    return button_rect

def reset_grid():
    global grid
    grid = [row[:] for row in grid_original]

def solve_sudoku(board):
    empty = find_empty_location(board)
    if not empty:
        return True
    row, col = empty
    for num in range(1, 10):
        if is_valid(board, row, col, num):
            board[row][col] = num
            if solve_sudoku(board):
                return True
            board[row][col] = 0
    return False

def find_empty_location(board):
    for i in range(9):
        for j in range(9):
            if board[i][j] == 0:
                return (i, j)
    return None

def draw_buttons(win):
    global reset_button_rect, solve_button_rect
    reset_button_rect = create_button(win, "Reset", (50, 550), (150, 50))
    solve_button_rect = create_button(win, "Solve", (250, 550), (150, 50))

def main():
    global grid_original, grid
    difficulty = get_difficulty()
    grid_original = select_board(difficulty)
    grid = [row[:] for row in grid_original]

    pygame.init()
    win = pygame.display.set_mode((WIDTH, WIDTH))
    pygame.display.set_caption("Sudoku")
    win.fill(background_color)
    pygame.draw.rect(win, (0, 0, 0), (50, 50, 450, 450), 2)
    draw_grid(win)
    draw_numbers(win)
    
    # Create buttons
    draw_buttons(win)
    
    pygame.display.update()

    while True:
        for event in pygame.event.get():
            if event.type == pygame.MOUSEBUTTONUP and event.button == 1:
                pos = pygame.mouse.get_pos()
                if reset_button_rect.collidepoint(pos):
                    reset_grid()
                    win.fill(background_color)
                    pygame.draw.rect(win, (0, 0, 0), (50, 50, 450, 450), 2)
                    draw_grid(win)
                    draw_numbers(win)
                    draw_buttons(win)
                    pygame.display.update()
                elif solve_button_rect.collidepoint(pos):
                    if solve_sudoku(grid):
                        win.fill(background_color)
                        pygame.draw.rect(win, (0, 0, 0), (50, 50, 450, 450), 2)
                        draw_grid(win)
                        draw_numbers(win)
                        draw_buttons(win)
                        pygame.display.update()
                    else:
                        show_error_message(win, pygame.font.SysFont('Comic Sans MS', 35))
                else:
                    insert(win, (pos[0]//50, pos[1]//50))

            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()

if __name__ == "__main__":
    main()
```
